```python
import pandas as pd
import os
from datetime import datetime
```

```python
SUPPORTED_STOCKS = [
    "AAPL",
    "MSFT",
    "GOOGL",
    "AMZN",
    "NVDA"
]

RESULTS_DIR = "../results"
REPORTS_DIR = "../reports"

os.makedirs(RESULTS_DIR, exist_ok=True)

all_ai_recommendations = []
```

```python
for ticker in SUPPORTED_STOCKS:

    print("=" * 70)
    print(f"Processing {ticker}")
    print("=" * 70)

    # -----------------------------
    # Load datasets
    # -----------------------------

    signals = pd.read_csv(
        f"{RESULTS_DIR}/{ticker}_recommendation.csv"
    )

    sentiment = pd.read_csv(
        f"{REPORTS_DIR}/{ticker}/news_sentiment.csv"
    )

    forecast = pd.read_csv(
        f"{REPORTS_DIR}/{ticker}/prophet_forecast.csv"
    )

    print("Recommendation")
    display(signals)

    print("Sentiment Sample")
    display(sentiment.head())

    print("Forecast Sample")
    display(forecast.head())


    # -----------------------------
    # Extract values
    # -----------------------------

    recommendation = signals.loc[0, "Recommendation"]

    confidence = signals.loc[0, "Confidence (%)"]

    current_price = signals.loc[0, "Current Price"]

    predicted_price = signals.loc[0, "Predicted Price"]

    trading_score = signals.loc[0, "Trading Score"]

    sentiment_score = signals.loc[0, "Average Sentiment"]


    # -----------------------------
    # Generate AI Insights
    # -----------------------------

    insights = []

    if predicted_price > current_price:

        insights.append(
            "AI predicts a potential upward movement in stock price."
        )

    else:

        insights.append(
            "AI predicts a possible downward movement in stock price."
        )


    if recommendation == "BUY":

        insights.append(
            "Technical indicators suggest a BUY opportunity."
        )

    elif recommendation == "SELL":

        insights.append(
            "Technical indicators suggest a SELL opportunity."
        )

    else:

        insights.append(
            "Current market conditions suggest HOLD."
        )


    if sentiment_score > 0:

        insights.append(
            "Positive news sentiment may support price growth."
        )

    elif sentiment_score < 0:

        insights.append(
            "Negative news sentiment may affect short-term performance."
        )

    else:

        insights.append(
            "Neutral news sentiment observed."
        )


    if trading_score >= 70:

        insights.append(
            "High confidence trading opportunity."
        )

    elif trading_score >= 50:

        insights.append(
            "Moderate confidence trading opportunity."
        )

    else:

        insights.append(
            "Low confidence. Trade carefully."
        )


    insights.append(
        f"Overall Recommendation: {recommendation} ({confidence:.2f}% Confidence)"
    )


    print("\nGenerated Insights\n")

    for i, text in enumerate(insights, start=1):

        print(f"{i}. {text}")


    recommendation_df = pd.DataFrame({

        "Ticker":[ticker],

        "Generated On":[datetime.now()],

        "Recommendation":[recommendation],

        "Confidence (%)":[confidence],

        "Trading Score":[trading_score],

        "Average Sentiment":[sentiment_score],

        "Insights":[" | ".join(insights)]

    })


    recommendation_df.to_csv(

        f"{RESULTS_DIR}/{ticker}_ai_recommendation.csv",

        index=False

    )

    all_ai_recommendations.append(recommendation_df)
```

```python
final_df = pd.concat(
    all_ai_recommendations,
    ignore_index=True
)

final_df.to_csv(
    f"{RESULTS_DIR}/all_ai_recommendations.csv",
    index=False
)

print("=" * 70)
print("LAB 11 COMPLETED SUCCESSFULLY")
print("=" * 70)

display(final_df)

print("\nGenerated Files\n")

for ticker in SUPPORTED_STOCKS:

    print(f"{ticker}_ai_recommendation.csv")

print("\nCombined File")

print("all_ai_recommendations.csv")
```
