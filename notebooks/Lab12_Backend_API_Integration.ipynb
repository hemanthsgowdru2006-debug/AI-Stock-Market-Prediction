```python
from flask import Flask, jsonify
import pandas as pd
from pathlib import Path
```

```python
PROJECT_ROOT = Path.cwd().parent

DATA_DIR = PROJECT_ROOT / "data"
REPORTS_DIR = PROJECT_ROOT / "reports"
RESULTS_DIR = PROJECT_ROOT / "results"

SUPPORTED_STOCKS = [
    "AAPL",
    "MSFT",
    "GOOGL",
    "AMZN",
    "NVDA"
]

print("Project Root:", PROJECT_ROOT)
```

```python
stock_data = {}
forecast_data = {}
sentiment_data = {}
signal_data = {}
recommendation_data = {}

for ticker in SUPPORTED_STOCKS:

    stock_data[ticker] = pd.read_csv(
        DATA_DIR /
        ticker /
        "processed" /
        "processed_stock_data.csv"
    )

    forecast_data[ticker] = pd.read_csv(
        REPORTS_DIR /
        ticker /
        "prophet_forecast.csv"
    )

    sentiment_data[ticker] = pd.read_csv(
        REPORTS_DIR /
        ticker /
        "news_sentiment.csv"
    )

    signal_data[ticker] = pd.read_csv(
        RESULTS_DIR /
        f"{ticker}_recommendation.csv"
    )

    recommendation_data[ticker] = pd.read_csv(
        RESULTS_DIR /
        f"{ticker}_ai_recommendation.csv"
    )

print("All datasets loaded successfully.")
```

```python
app = Flask(__name__)

print("Flask App Created Successfully")
```

```python
@app.route("/")
def home():

    return jsonify({

        "Project":"AI Stock Market Prediction Dashboard",

        "Version":"2.0",

        "Supported Stocks":SUPPORTED_STOCKS,

        "Example":

        "/stock/AAPL",

        "Available APIs":[

            "/stock/<ticker>",

            "/prediction/<ticker>",

            "/sentiment/<ticker>",

            "/signal/<ticker>",

            "/recommendation/<ticker>"

        ]

    })
```

```python
@app.route("/stock/<ticker>")
def stock_api(ticker):

    ticker = ticker.upper()

    if ticker not in SUPPORTED_STOCKS:

        return jsonify({"Error":"Invalid Stock"})

    return jsonify(

        stock_data[ticker]

        .tail(30)

        .to_dict(

            orient="records"

        )

    )
```

```python
@app.route("/prediction/<ticker>")
def prediction_api(ticker):

    ticker = ticker.upper()

    if ticker not in SUPPORTED_STOCKS:

        return jsonify({"Error":"Invalid Stock"})

    return jsonify(

        forecast_data[ticker]

        .to_dict(

            orient="records"

        )

    )
```

```python
@app.route("/sentiment/<ticker>")
def sentiment_api(ticker):

    ticker = ticker.upper()

    if ticker not in SUPPORTED_STOCKS:

        return jsonify({"Error":"Invalid Stock"})

    return jsonify(

        sentiment_data[ticker]

        .to_dict(

            orient="records"

        )

    )
```

```python
@app.route("/signal/<ticker>")
def signal_api(ticker):

    ticker = ticker.upper()

    if ticker not in SUPPORTED_STOCKS:

        return jsonify({"Error":"Invalid Stock"})

    return jsonify(

        signal_data[ticker]

        .to_dict(

            orient="records"

        )

    )
```

```python
@app.route("/recommendation/<ticker>")
def recommendation_api(ticker):

    ticker = ticker.upper()

    if ticker not in SUPPORTED_STOCKS:

        return jsonify({"Error":"Invalid Stock"})

    return jsonify(

        recommendation_data[ticker]

        .to_dict(

            orient="records"

        )

    )
```

```python
if __name__ == "__main__":

    app.run(
    host="127.0.0.1",
    port=5000,
    debug=False,
    use_reloader=False
) 
```
