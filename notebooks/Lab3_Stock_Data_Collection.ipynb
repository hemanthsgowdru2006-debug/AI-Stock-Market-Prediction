```python
import os
import requests
import pandas as pd
import yfinance as yf
```

```python
import os

SUPPORTED_STOCKS = [
    "AAPL",
    "MSFT",
    "GOOGL",
    "AMZN",
    "NVDA"
]

TICKER = "NVDA"      # Change only this line when generating another stock

RAW_DIR = f"../data/{TICKER}/raw"

os.makedirs(RAW_DIR, exist_ok=True)

print("Supported Stocks:", SUPPORTED_STOCKS)
print("Current Stock:", TICKER)
print("Folder:", RAW_DIR)
```

```python

# Replace with your Alpha Vantage API key
ALPHA_API_KEY = "T79N9JI77ZRXZQGZ"
```

```python
stock = yf.download(
    TICKER,
    start="2023-01-01",
    auto_adjust=False,
    progress=False
)

# Flatten MultiIndex columns if present
if isinstance(stock.columns, pd.MultiIndex):
    stock.columns = stock.columns.get_level_values(0)

stock.reset_index(inplace=True)

stock.to_csv(f"{RAW_DIR}/yahoo_stock.csv", index=False)
```

```python
print("=" * 60)
print("YAHOO FINANCE DATASET")
print("=" * 60)

print(stock.info())

print("\nMissing Values\n")

print(stock.isnull().sum())

print("\nRows :", len(stock))
```

```python
stock.to_csv(
    f"{RAW_DIR}/yahoo_stock.csv",
    index=False
)

print("Stock data saved successfully.")
```

```python
url = (
    "https://www.alphavantage.co/query"
    "?function=OVERVIEW"
    f"&symbol={TICKER}"
    f"&apikey={ALPHA_API_KEY}"
)

response = requests.get(url)

company = response.json()

if "Symbol" in company:
    print("Company information downloaded successfully.")
else:
    print("Warning: Could not retrieve company information.")
```

```python
company_df = pd.DataFrame([company])

company_df.head()
```

```python
company_df.to_csv(
    f"{RAW_DIR}/company_info.csv",
    index=False
)

print("Company information saved successfully.")
```
